### Instructions for how to save and load a model

```python
import os
import torch
from microns.main import scan
from fnn.microns.load import units
from fnn.microns.build import network
```

```python
# test model 
# look at data/microns/scans.csv for numbers that work for session and scan_idx
session = 4
scan_idx = 7
pred_model, table = scan(session, scan_idx, directory = os.path.join(os.getcwd(), "data","microns")) 
```

Save Model

```python
# save modelname.sate_dict() as follows, save name as .pt file 
# can specify directory in name if needed
torch.save(pred_model.state_dict(), "pred_model1.pt")
```

Load Model


```python
# make sure to instantiate new Visual model

dir = os.path.join(os.getcwd(), "fnn", "data", "microns")
pred_model2 = network(units(session, scan_idx, dir))
pred_model2.load_state_dict(torch.load("pred_model1.pt", weights_only = True))
pred_model2.eval()
```
